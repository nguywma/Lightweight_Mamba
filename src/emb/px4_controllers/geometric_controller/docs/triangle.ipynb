```python
import matplotlib.pyplot as plt
import numpy as np
snap_max = 10
dt = 0.01
num_steps = 400

jerk = 0
acceleration = 0
velocity = 0
position = 0

times = dt * np.arange(num_steps)

jerk_values = []
acceleration_values = []
velocity_values = []
position_values = []
slot_values = [1,-1,-1,1,-1,1,1,-1,0]
for i in range(num_steps):
    slot = int(i // 50)
    jerk += snap_max * slot_values[slot] * dt
    acceleration += jerk * dt
    velocity += acceleration * dt
    position += velocity * dt
    
    jerk_values.append(jerk)
    acceleration_values.append(acceleration)
    velocity_values.append(velocity)
    position_values.append(position)

fig, ax = plt.subplots(5, 1, figsize=(10, 20))
fig.subplots_adjust(hspace=0.5)

ax[0].plot(times, jerk_values)
ax[0].set_ylabel('Jerk')
ax[0].grid(True)

ax[1].plot(times, acceleration_values)
ax[1].set_ylabel('Acceleration')
ax[1].grid(True)

ax[2].plot(times, velocity_values)
ax[2].set_ylabel('Velocity')
ax[2].grid(True)

ax[3].plot(times, position_values)
ax[3].set_ylabel('Position')
ax[3].grid(True)

ax[4].plot(times, jerk_values, label='Jerk')
ax[4].plot(times, acceleration_values, label='Acceleration')
ax[4].plot(times, velocity_values, label='Velocity')
ax[4].plot(times, position_values, label='Position')
ax[4].set_xlabel('Time')
ax[4].set_ylabel('Value')
ax[4].grid(True)
ax[4].legend()

plt.show()
```

```python
import numpy as np
import math
import matplotlib.pyplot as plt
import matplotlib.animation as animation
from IPython.display import HTML
# Set up the figure and axis
fig, ax = plt.subplots()
ax.set_xlim(-1.5, 1.5)
ax.set_ylim(-1.5, 1.5)
ax.set_aspect('equal')

# Create a circle
theta = np.linspace(0, 2*np.pi, 100)
x = np.cos(theta)
y = np.sin(theta)
circle, = ax.plot(x, y)

# Create a point
point, = ax.plot([], [], 'ro')

# Initialize the data
t = 0
omega = math.pi/2
dt = 0.01
xdata, ydata = [], []

# Define the update function
def update(num):
    global t, omega, dt, xdata, ydata
    t += dt
    x = np.cos(omega*t)
    y = np.sin(omega*t)
    xdata.append(x)
    ydata.append(y)
    point.set_data(x, y)
    return point,

# Create the animation
ani = animation.FuncAnimation(fig, update, frames=200, interval=10, blit=True)
# Display the animation
HTML(ani.to_jshtml())
```
